### Log returns and cumulative growth
Log returns sum cumulatively into a running total. Exponentiating that cumulative sum converts it back into an actual growth multiple — a direct, concrete way to answer "how much would $1 invested at the start be worth now?"

```python
log_returns = np.log(data['Close']).diff()
log_returns.cumsum()
```

```python
# total growth multiple over the whole period
np.exp(log_returns.cumsum()).iloc[-1]
```

```python
log_returns.describe()
```
