# NumPy: Array Attributes & Creation
### OIM 7502 — NumPy Series, Part 2 of 7

Now that you've seen what an array is, let's look at how to inspect one and how to create arrays without typing out every value by hand.

### Core attributes
- `size` — total number of elements (rows × columns)
- `itemsize` — bytes used by a single value
- `shape` — a tuple giving the length of each dimension
- `ndim` — how many dimensions the array has

### Reshaping
`reshape` lets you change an array's dimensions without changing its data — the total number of elements has to stay the same. Passing `-1` for one dimension tells NumPy "figure this one out for me."

### Creating arrays without typing every value
`arange` and `linspace` are the two you'll reach for constantly:
- `np.arange(start, stop, step)` — like Python's `range`, but returns an array
- `np.linspace(start, stop, num)` — `num` evenly spaced points between `start` and `stop`, **inclusive** of both ends

### Special arrays
Useful for dimensioning an array before you know what values it will hold — for example, pre-allocating space you'll fill in with a loop or computation.

**A warning about `np.empty`:** it's faster than `zeros` because it skips initializing the values — but that means the values are unpredictable leftover memory, not zeros. Only use it when you're about to overwrite every element yourself.
