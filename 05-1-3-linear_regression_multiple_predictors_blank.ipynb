# Linear Regression: Multiple Predictors with statsmodels
### OIM 7502 — Linear Regression Series, Part 3 of 5

Area alone left 71% of price variation unexplained. Houses obviously vary in more than square footage — let's add bedrooms, bathrooms, and stories, and see how much that improves the fit.

### Before modeling: check correlations
A quick sanity check on which predictors are related to price — and to each other. Two predictors that are highly correlated with *each other* (not just with the target) can make coefficients unstable, the same multicollinearity issue we'll see again with logistic regression's dummy variables.

`bathrooms` (0.52) and `stories` (0.42) both correlate with price nearly as strongly as `area` (0.54) — worth including. The predictors correlate modestly with each other (all below 0.41), so multicollinearity isn't a major concern here.

### What changed
R² jumped from 0.287 (area alone) to **0.536** — the three added predictors explain a lot of price variation that area alone missed. Every coefficient is now a "holding everything else constant" statement: an extra bathroom is associated with +$1,200k... in price, *holding area, bedrooms, and stories fixed*. Note the `const` (intercept) is no longer significant (p=0.264) on its own — that's fine; we don't usually care whether the intercept itself is meaningful.

### How far off are the predictions, in real terms?
R² measures explained variance, but it doesn't tell you the size of a typical mistake in dollars. Root-mean-squared-error (RMSE) does — same units as `y` (thousands of dollars here).

On average, this model's price predictions are off by roughly $1.27 million (remember `y` is in thousands) — a useful number to have in hand before deciding whether this model is "good enough" for whatever it's being used for.
