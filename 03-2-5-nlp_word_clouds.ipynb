# NLP: Word Clouds
### OIM 7502 — NLP Series, Part 5 of 5

A word cloud is a quick, visual way to see which words dominate a body of text — size corresponds to frequency. It's more of a communication tool than an analytical one, but it's an easy, engaging way to summarize text for an audience.

Word clouds are a case where stopword removal is exactly the right call — you don't want a giant "THE" dominating the picture. `WordCloud` has its own built-in stopword list (`STOPWORDS`), separate from NLTK's.

**Series wrap-up:** across these five notebooks, the same scraped text went through four different transformations — raw extraction, tokenization/stopword-filtering, n-gram frequency counting, sentiment scoring, and visual summarization. The one rule that ties Parts 2–5 together: decide *which version* of the text (original vs. filtered) is right for the tool you're using, rather than reusing whatever variable happens to already exist.
