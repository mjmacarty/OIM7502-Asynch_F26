# NumPy: Indexing & Slicing
### OIM 7502 — NumPy Series, Part 3 of 7

This is where NumPy really starts to feel different from a plain Python list — especially boolean and fancy indexing, which you'll use constantly once you get to pandas.

### Basic slicing
For a 2D array, think of indexing as `array[rows, columns]`.

### Boolean indexing
You can index an array with a condition, and NumPy returns only the elements where that condition is `True`. This is the pattern you'll use over and over in pandas.

### Boolean indexing for assignment
You can use a boolean mask on the left side of an assignment to selectively update values.

### Fancy indexing
Pass a *list of indices* (instead of a slice) to pull out specific, non-contiguous elements in whatever order you want.

### `np.where`
`np.where(condition, if_true, if_false)` is a vectorized if/else — it's the NumPy equivalent of writing a loop with an `if` statement, but far faster.
