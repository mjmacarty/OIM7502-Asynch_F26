# NumPy: Working with Random Numbers

NumPy's `random` module can generate values from many distributions. We'll focus on the two you'll use constantly: uniform and normal.

### Reproducibility with a seed
Random means different results every time you run it — which makes demos and debugging painful. Setting a seed makes the "random" sequence repeatable, so you (and anyone following along) get the exact same numbers every run.

**Set the seed once, near the top of your notebook or script** — not before every single call — or you'll accidentally make every "random" draw after the first one identical too.

### Standard normal (mean 0, std 1)
`np.random.standard_normal([size])` takes a single optional argument for how many values you want.

### Uniform distributions
- `np.random.randn(size)` — standard normal, an alternate spelling of the function above
- `np.random.randint(low, high=None, size=None)` — random integers; note `high` is **excluded** (a half-open interval)

### Normal distribution with a chosen mean and spread
`loc` sets the mean, `scale` sets the standard deviation. If you don't specify them, you get the standard normal (mean 0, std 1) — same as `standard_normal`.
