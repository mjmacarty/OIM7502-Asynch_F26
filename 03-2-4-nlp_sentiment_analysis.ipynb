# NLP: Sentiment Analysis with VADER
### OIM 7502 — NLP Series, Part 4 of 5

VADER (Valence Aware Dictionary and sEntiment Reasoner) is a lexicon-based sentiment tool built into NLTK. It scores text on four dimensions: negative, neutral, positive, and a **compound** score from -1 (most negative) to +1 (most positive) that's usually the one you actually care about.

### Important: run VADER on the ORIGINAL text, not stopword-stripped text
This is the single biggest mistake to avoid in this series. VADER is specifically designed to read negation ("not good"), degree modifiers ("very good"), capitalization, and punctuation as part of the sentiment signal — all things a naive stopword-removal pass can quietly destroy.

### A caveat that applies regardless of preprocessing
Even scored correctly, VADER (like most lexicon-based tools) can't reliably detect sarcasm or backhanded phrasing — "great, another Monday" reads as positive to VADER because of the word "great," tone notwithstanding. Treat a VADER score as a useful signal, not a verdict, especially on informal or sarcastic text.
