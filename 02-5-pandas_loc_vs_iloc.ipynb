# pandas: `loc` vs. `iloc` Row Slicing
### OIM 7502 — pandas Series, Part 5 of 9

Both slice rows, but they answer different questions:
- **`.loc`** — "give me rows *by label*" (inclusive of both endpoints when slicing)
- **`.iloc`** — "give me rows *by position*" (exclusive of the end position, like Python list slicing)

These happened to line up here because we already knew which positions corresponded to that date range. In general they won't match unless you check first — don't assume a `.loc` label range and an `.iloc` position range of "the same size" give you the same rows.
