## 1. Headers

Headers use `#` symbols — one `#` is the biggest, six is the smallest:

```
# Header 1
## Header 2
### Header 3
```

Use headers to organize a notebook the way you'd organize a report — one `#` for the notebook's title, `##` for major sections, `###` for subsections. Don't skip levels.

**Practice:** In this cell, add a level-1 header with your name, and a level-2 header titled "About This Notebook."

## 2. Emphasis (bold, italic, etc.)

```
*italic text* or _italic text_
**bold text** or __bold text__
***bold and italic***
~~strikethrough~~
```

**Practice:** Write one paragraph of italic text, and one sentence containing bold text.

## 3. Lists

**Unordered** (use `-`, `*`, or `+` — pick one and stay consistent):
```
- First item
- Second item
  - Nested item
```

**Ordered:**
```
1. First step
2. Second step
```

**Practice:** Write a bulleted list of three things you want to learn this term.

**Practice:** Write a numbered list of the three steps in your data analysis workflow.

## 4. Links and Images

```
[link text](https://example.com)
![alt text](path/to/image.png)
```

The only difference between a link and an image is the `!` in front.

**Practice:** Add a link to a website you use for reference (e.g. the pandas or NumPy docs).

## 5. Code

**Inline code** uses single backticks: `` `like this` ``.

**Code blocks** use triple backticks, optionally with a language name for syntax highlighting:

````
```python
import pandas as pd
```
````

**Practice:** Add a fenced code block containing a single line of Python.

## 6. Blockquotes

```
> This is a blockquote.
> It can span multiple lines.
```

Useful for calling out a quote, a warning, or a note you want visually separated from the surrounding text.

**Practice:** Add a blockquote containing a quote or a note to yourself.

## 7. Tables

```
| Column A | Column B |
|----------|----------|
| 1        | 2        |
```

The dashes in the second row tell Markdown "this is a table header" — they need at least three dashes per column.

**Practice:** Add a table with at least 2 columns and 3 rows.

## 8. Horizontal Rules

Three or more dashes, asterisks, or underscores on their own line:

```
---
```

Useful for visually separating sections — this whole notebook uses them.

**Practice:** Add a horizontal rule separating two sections.

## 9. Line Breaks (the one that trips everyone up)

A single line break in your raw markdown is **ignored** when rendered — to force a line break, end a line with two trailing spaces, or leave a fully blank line to start a new paragraph.

**Practice:** Write two lines of text that should appear on separate lines when rendered, using a forced line break (two trailing spaces) between them.
