# APIs: Convenience Libraries (yfinance)
### OIM 7502 — API Series, Part 3 of 4

Writing raw `requests` calls and parsing JSON by hand works everywhere, but plenty of popular APIs have a Python package built specifically for them — a **convenience library**. It handles the endpoint URLs, auth, and JSON parsing for you, and typically hands the result back already wrapped in a pandas Series or DataFrame.

`yfinance` (Yahoo Finance) is a good example: it's essentially a purpose-built wrapper that does everything the last two notebooks did manually, in one function call.

Notice what you got back directly: a DataFrame, already indexed by date, with the columns you'd expect (Open/High/Low/Close/Volume) — no `requests.get`, no `json.loads`, no manual parsing. That's the entire value proposition of a convenience library.
