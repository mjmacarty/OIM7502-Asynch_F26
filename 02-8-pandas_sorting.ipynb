# pandas: Sorting Data
### OIM 7502 — pandas Series, Part 8 of 9

### Sorting by index
`sort_index()` sorts by the row (or column) labels, not by any data values.

### Sorting by values
`sort_values(by=...)` sorts by the actual data in a column.

### Sorting columns instead of rows
`axis=1` applies the same idea to column labels — here, alphabetically.
