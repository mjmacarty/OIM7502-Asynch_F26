# Logistic Regression: Evaluation with a Confusion Matrix
### OIM 7502 — Logistic Regression Series, Part 4 of 4

A fitted classifier's predictions are probabilities, and `statsmodels` defaults to a 0.5 cutoff to turn those into a yes/no call. A **confusion matrix** lays out exactly where those calls went right and wrong — building on the reduced model from Part 3.

### Building the confusion matrix
`.pred_table()` gives you a 2x2 array of [[true negatives, false positives], [false negatives, true positives]] at the default 0.5 cutoff. Wrapping it in a labeled DataFrame makes it far easier to read.

### From four numbers to the metrics that matter
Every classification metric you'll see elsewhere (in `sklearn`, in a paper, in a business dashboard) is built from these same four counts.

### Why accuracy alone is misleading here
Accuracy looks solid (~72%) — but recall (TPR) is only ~35%. The model is missing nearly two-thirds of the actual affairs, mostly because the classes are imbalanced (68% "no affair") and a 0.5 cutoff leans heavily toward predicting the majority class. This is exactly the imbalanced-target situation flagged back in Part 1 — always check recall/precision alongside accuracy, not instead of it, whenever your classes aren't roughly 50/50.
