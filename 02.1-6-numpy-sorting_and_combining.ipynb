# NumPy: Sorting & Combining Arrays
### OIM 7502 — NumPy Series, Part 6 of 7

### Sorting
`np.sort` returns a sorted copy. `np.argsort` returns the *indices* that would sort the array — useful when you need to sort one array based on the order of another (e.g., sort names by score).

### Combining arrays
`concatenate` joins arrays along an existing axis. `vstack`/`hstack` are convenience wrappers for the common 2D cases: stacking vertically (adding rows) or horizontally (adding columns).

### Transpose
`.T` swaps rows and columns — handy when your data is oriented the wrong way for the operation you want to do next.
