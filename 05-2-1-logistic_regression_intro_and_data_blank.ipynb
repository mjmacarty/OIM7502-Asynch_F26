# Logistic Regression: Intro & the Data
### OIM 7502 — Logistic Regression Series, Part 1 of 4

Linear regression predicts a continuous number. Logistic regression predicts a **probability** of a binary outcome (yes/no, 0/1) — it's the natural next step once your target stops being a number and becomes a category.

We'll use a classic teaching dataset that ships directly with `statsmodels`: Fair's 1978 "affairs" dataset, which surveyed married people about their marriage and whether they'd had an extramarital affair. No CSV to track down — it loads straight from the library.

### The columns
Each row is one respondent: marriage-satisfaction rating, age, years married, number of children, religiousness, education, occupation codes for both spouses, and `affairs` — a continuous measure of how many affairs they reported (0 if none).

### From a continuous measure to a binary target
`affairs` is continuous, but the question we actually want to answer is binary: **did this person have an affair at all?** We convert `affairs` into a 0/1 flag.

About 32% of respondents (2,053 of 6,366) reported at least one affair. That's a meaningfully imbalanced target (not 50/50) — worth remembering when we get to evaluation in Part 4, since accuracy alone can be misleading on imbalanced classes.

### A first look: does anything differ by group?
Before fitting any model, compare the average of every column between the two groups — a quick, model-free way to spot which variables might matter.

A few things jump out even before modeling: the affair group reports lower marriage satisfaction (`rate_marriage`), more years married, and slightly lower religiousness. That's exactly the kind of pattern logistic regression will let us quantify formally — how much does each variable actually move the odds, holding the others constant?
