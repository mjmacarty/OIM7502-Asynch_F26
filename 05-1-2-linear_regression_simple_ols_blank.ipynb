Ordinary least squares (OLS) has implementations across the Python ecosystem — NumPy (seen earlier, from scratch), SciPy, `statsmodels`, and `sklearn`. `statsmodels` is the one that reads most like a statistics textbook: it hands back p-values, confidence intervals, and R² alongside the coefficients, which is why we start there.

A clear upward trend, with real scatter around it — exactly the kind of relationship linear regression is built to summarize.

### Adding the intercept term
`statsmodels` doesn't add an intercept automatically — you have to ask for it explicitly with `sm.add_constant`. Skip this and the model is forced through the origin (price = 0 when area = 0), which almost never makes sense.

- **R-squared (0.287)**: area alone explains about 29% of the variation in price — real, but far from the whole story (bedrooms/bathrooms/stories are still on the table).
- **coef for `area` (0.462)**: each additional square foot is associated with roughly $462 more in price, holding nothing else constant (there's nothing else in this model yet).
- **P>|t| for `area` (0.000)**: the relationship is statistically significant — essentially no chance this slope is due to noise alone.
