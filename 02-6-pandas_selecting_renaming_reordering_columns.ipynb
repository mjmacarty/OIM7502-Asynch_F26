# pandas: Selecting, Renaming & Reordering Columns
### OIM 7502 — pandas Series, Part 6 of 9

### Selecting one column
Dot notation and bracket notation do the same thing for a simple column name — but dot notation breaks if the column name has spaces or matches a DataFrame method name, so bracket notation is the safer default habit.

### Selecting multiple columns
Pass a *list* of column names — note the double brackets: the outer one is the indexing operation, the inner one is the list itself.

### Renaming columns
`rename` returns a **new** DataFrame by default — it does not modify the original unless you pass `inplace=True` or reassign the result.

### Reordering columns
Just index with the columns in the order you want them.
