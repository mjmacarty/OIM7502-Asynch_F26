# NumPy: Broadcasting
### OIM 7502 — NumPy Series, Part 5 of 7

Broadcasting is how NumPy handles operations between arrays of *different* shapes without you writing a loop. You've already seen the simplest case — adding a scalar to an array. It generalizes: NumPy will "stretch" a smaller array across a larger one, as long as their shapes are compatible.

### Broadcasting a 1D array against a 2D array
This is the more useful case: a per-column adjustment, applied to every row at once.

Here, `curve` has shape `(3,)` and `grades` has shape `(3, 3)`. NumPy lines `curve` up against the **last** dimension of `grades` (the columns) and repeats it down every row. That's the general rule: shapes are compared from the right, and a dimension of size 1 (or a missing dimension) gets stretched to match.

### When broadcasting fails
If the shapes genuinely don't line up, NumPy raises an error instead of guessing what you meant.

**Reading the error:** `(3,3) (2,)` tells you exactly what didn't line up — 3 columns on the left, 2 elements on the right. When you hit a broadcasting error, the fix is almost always to check `.shape` on both arrays and figure out which dimension you expected to match but didn't.
