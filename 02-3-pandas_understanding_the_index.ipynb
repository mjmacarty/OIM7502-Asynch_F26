This is the fix — and it's a step you'll need constantly whenever you load date columns from a CSV, since `read_csv` never infers dates for you automatically.

**Takeaway:** dates that *look* right when printed are not the same as dates pandas can actually compute with. Always convert explicitly with `pd.to_datetime()` or `pd.DatetimeIndex()` rather than assuming `read_csv` handled it — it didn't, unless you told it to with `parse_dates=`.
