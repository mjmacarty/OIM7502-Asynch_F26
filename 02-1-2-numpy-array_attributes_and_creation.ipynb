NumPy is the foundation of the Python data science ecosystem — pandas, scikit-learn, and most of the libraries we use are built on top of it. It's a huge library; we're going to hit the highlights that matter for the rest of this course.

**What makes a NumPy array different from a Python list?**
- Optimized for numeric computation — much faster than lists for the same operations
- Can be any number of dimensions (we'll mostly work with 1D and 2D)
- Every element must be the **same data type** (a list can mix types freely)
- You "dimension" an array when you create it, and reshape it explicitly when you want to change that

### The `+` operator means something completely different for each

Note that `a + 1` didn't change `a` itself — like most operations, it returns a *new* array. If you want to update `a`, you need to reassign it.

### The single-dtype restriction
Every element in a NumPy array shares one data type. If you hand it mixed types, NumPy will **upcast** everything to the most general type it can — this can bite you if you're not expecting it.

### Why bother? Performance.
NumPy operations run in compiled C code under the hood, not in the Python interpreter loop. On a large array, that difference is dramatic.

**Takeaway:** for anything numeric and at scale, reach for NumPy arrays over lists. The syntax is similar, but the performance and the vectorized (elementwise) behavior are what make it worth learning.
