### Boolean filtering (a deeper look)
You've seen `array[condition]` in the NumPy series — it works exactly the same way on a DataFrame, and it's the single most common way you'll subset real data.

### `value_counts`
The go-to method for summarizing a categorical column — counts of each unique value, sorted descending by default.

### GroupBy basics
`groupby` splits the data into groups by a column's values, then applies an aggregation to each group separately — this is the pattern behind almost every "average X by category" question you'll ask of real data.

You can group by more than one column at once, and pick which columns to aggregate rather than taking all numeric ones.
