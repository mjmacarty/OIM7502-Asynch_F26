We'll pull real text from a live page and work with it through the rest of this series.

```python
from bs4 import BeautifulSoup as bs  # no, the pun is not lost
import nltk
import requests
```

```python
url = "https://alphabench.com/data/python-loan-analysis.html"
html = requests.get(url).text
```

### Parsing HTML
Wrapping the raw HTML in a `BeautifulSoup` object gives you a searchable, navigable tree instead of one giant string.

```python
soup = bs(html, 'html.parser')
```

### Getting text out of tags
`.p` grabs the *first* `<p>` tag; `.text` strips the surrounding HTML and gives you just the readable content.

```python
soup.p.text
```

### Extracting all hyperlinks
`.find_all('a')` returns every anchor tag on the page. Each one has both visible text (`.text`) and a destination (`.get('href')`).

```python
for link in soup.find_all('a'):
    print(link.text)
```

```python
for link in soup.find_all('a'):
    print(link.get('href'))
```

### Pulling all paragraph text into one string
For the rest of this series, we want the full body text, not just the first paragraph — concatenate every `<p>` tag's text together.

```python
text = ""
for paragraph in soup.find_all('p'):
    text += paragraph.text

len(text)
```
