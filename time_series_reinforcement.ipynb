# Reinforcement Activity: Time Series Basics & Forecasting

In this activity you'll practice working with a DatetimeIndex, resampling, and comparing two simple forecasting methods: Simple Moving Average (SMA) and Simple Exponential Smoothing (SES).

**A note on statsmodels:** in current versions, `SimpleExpSmoothing` requires you to pass `initialization_method` explicitly (e.g. `initialization_method="legacy-heuristic"`) — older tutorials online may not show this and will error without it.

**A note on resample frequency codes:** pandas changed its resample aliases between versions — `'M'` (month-end) became `'ME'`, and `'BQ'` became `'BQE'`, starting around pandas 2.2. If you get an `Invalid frequency` error on a resample call, try the other version of the code (`'M'` vs `'ME'`) — it depends on which pandas version you have installed.

### 1. Generate a synthetic daily time series
Create a `pd.date_range` covering all of 2026, and generate a random daily value (e.g. simulated units sold) for each date using NumPy. Set a random seed first so your results are reproducible. Build a DataFrame with the date range as a proper `DatetimeIndex`.

### 2. Slice and inspect
Using `.loc`, slice out just the month of June, and separately slice a date range covering Q3 (July–September). Add columns for day of week and month number using the datetime index accessors.

### 3. Resample
Resample your daily series to **weekly** totals and to **monthly** averages. Briefly comment (markdown) on what changes about the data at each frequency, and when you'd choose one over the other.

### 4. Simple Moving Average (SMA)
Compute a 7-day simple moving average of your series. Shift it by 1 so it represents a genuine one-step-ahead forecast (i.e., today's SMA forecast should only use data through yesterday).

### 5. Simple Exponential Smoothing (SES)
Using `statsmodels`, fit a `SimpleExpSmoothing` model on your series with `smoothing_level=0.3` (don't let it optimize). Extract the fitted values, shifted by 1 for the same reason as above.

### 6. Evaluate and compare
Drop rows with missing values (from the shift), then compute MSE, MAD, and MAPE for both the SMA and SES forecasts against the actual values. Which method performed better on your data? Briefly comment on why that might be, given how your synthetic data was generated.

### 7. Plot
Plot the actual series alongside both forecasts (SMA and SES) on the same chart, well labeled.
