### Dropping columns or rows
Same rule as `rename` — `.drop()` returns a new object by default; use `inplace=True` or reassign to actually keep the change.

### Detecting missing data
`.isna()` (or `.isnull()`, an alias) flags missing values; `.sum()` on top of that gives you a quick per-column count.

Real financial data is rarely this clean — let's manufacture some gaps to see the tools in action.

### Handling missing data
Three common strategies, in roughly increasing order of sophistication:
- **`dropna()`** — remove rows (or columns) with missing values entirely
- **`fillna(value)`** — fill with a fixed value
- **`fillna(method=...)`** — fill using nearby data (forward-fill / backward-fill)

**Which one is right depends on the data.** For a stock price series, forward-filling (assume the price held steady) usually makes more sense than filling with the overall mean — but that's a judgment call you have to make deliberately, not a default to reach for blindly.
