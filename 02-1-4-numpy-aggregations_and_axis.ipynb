# NumPy: Aggregations & the `axis` Parameter
### OIM 7502 — NumPy Series, Part 4 of 7

### Aggregation functions
`sum`, `mean`, `min`, `max`, `std` all work the way you'd expect on a 1D array.

### The `axis` parameter
On a 2D array, you have to tell NumPy *which direction* to aggregate across. This trips almost everyone up the first time:
- `axis=0` collapses **down the rows** — you get one result per **column**
- `axis=1` collapses **across the columns** — you get one result per **row**

A quick trick if you're ever unsure which axis you want: `axis=0` acts *down* (fewer rows survive — one result per column); `axis=1` acts *across* (fewer columns survive — one result per row).
