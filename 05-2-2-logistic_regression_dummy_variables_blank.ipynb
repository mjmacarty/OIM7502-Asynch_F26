# Logistic Regression: Dummy Variables & the Design Matrix
### OIM 7502 — Logistic Regression Series, Part 2 of 4

`occupation` and `occupation_husb` are stored as numbers (1-6), but they're really **categories**, not a quantity — occupation "4" isn't twice occupation "2." Feeding them into a model as raw numbers would falsely imply an ordering/scale that isn't there. The fix: turn each category into its own 0/1 **dummy variable**.

`pd.get_dummies` returns `bool` columns in current pandas — the explicit `.astype(int)` gives you 0/1 integers instead, which is what `statsmodels` expects.

### The multicollinearity trap: drop one dummy per category
This is the step that's easy to skip and breaks everything downstream. If you keep *all six* occupation dummies, they sum to 1 in every row — which means one is always exactly predictable from the other five. That's perfect multicollinearity, and it makes the model's coefficients unstable or meaningless.

The fix: drop **one** dummy from each category to serve as the baseline. Every remaining dummy's coefficient is then interpreted "relative to that baseline" (here, occupation 1).
