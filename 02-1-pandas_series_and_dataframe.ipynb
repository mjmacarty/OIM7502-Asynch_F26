# pandas: Series & DataFrame Basics
### OIM 7502 — pandas Series, Part 1 of 9

pandas is the premier data science library in the Python ecosystem — a wrapper around, among other things, NumPy and matplotlib. The name comes from "panel data." It has two core objects:

- **Series** — a vector of data with an index
- **DataFrame** — a collection of Series sharing a single index (the columns are also an index)

(A bit of trivia: pandas was originally developed by a quant at a hedge fund, who was semi-retired from full-time development by his late 20s.)
