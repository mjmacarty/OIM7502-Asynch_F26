# Linear Regression: The Same Model in scikit-learn
### OIM 7502 — Linear Regression Series, Part 4 of 5

`statsmodels` is built for *inference* — understanding relationships, with p-values and confidence intervals front and center. `sklearn` is built for *prediction* — a consistent `.fit()`/`.predict()` API shared across dozens of model types, with no p-values by default. Same underlying math; different priorities. We'll refit the exact same model from Part 2 and confirm the numbers match.

Notice: no `sm.add_constant` here — `sklearn`'s `LinearRegression` adds the intercept for you automatically (that's `fit_intercept=True` by default).

Compare to Part 2's `multi_model.params`: `area` 0.3800, `bedrooms` 198.30, `bathrooms` 1200.18, `stories` 531.60, intercept -282.09. Same coefficients — `sklearn` and `statsmodels` are solving the identical least-squares problem, just exposing it through a different API.

**The takeaway isn't the numbers (they're identical) — it's the trade-off.** Reach for `statsmodels` when the question is "which variables matter, and how confident are we?" Reach for `sklearn` when the question is "build the best predictive pipeline" (it has train/test splitting, cross-validation, and dozens of other model types built around the same `.fit()`/`.predict()` pattern).
