# Linear Regression: Feature Scaling
### OIM 7502 — Linear Regression Series, Part 5 of 5

`area` ranges into the thousands; `bedrooms` and `stories` range from 1 to ~6. Plain OLS handles that fine (it fit just as well in Part 2) — but plenty of other techniques (regularized regression, KNN, gradient descent-based models, neural networks) are sensitive to features being on wildly different scales. `sklearn.preprocessing` gives you a few standard ways to rescale.

### StandardScaler: mean 0, standard deviation 1
Subtracts the mean and divides by the standard deviation for every column. The result is centered around 0, with units of "standard deviations from the mean" instead of the original units.

### MinMaxScaler: squeeze everything into [0, 1]
Instead of standard deviations, this rescales so the minimum value in the column becomes 0 and the maximum becomes 1 — useful when you want a bounded range rather than a specific distribution shape.

### MaxAbsScaler: divide by the largest absolute value
Scales so every value falls in [-1, 1] by dividing by the column's maximum absolute value. Handy for data that's already centered around 0 (it won't re-center anything), and it doesn't get thrown off by a single large value the way `MinMaxScaler` can.

**A common mistake worth calling out explicitly:** every one of these scalers returns the *transformed array* from `.fit_transform(...)` — index into that returned array (`price_maxabs[:10]`, as above), not into the scaler object itself. `max_abs[:10]` would raise a `TypeError`, because the fitted `MaxAbsScaler` object holds the scaling parameters, not the data.

### When does this actually matter?
Plain OLS (Parts 1-3) doesn't need scaling — it's invariant to it; the coefficients just rescale to compensate, and R²/predictions come out identical either way. Scaling starts to matter once you move to models that use distance directly (KNN, clustering) or gradient-based optimization (neural networks, many `sklearn` solvers) — a preview of things to come later in the course.
