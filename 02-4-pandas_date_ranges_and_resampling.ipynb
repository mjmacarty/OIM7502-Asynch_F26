# pandas: Date Ranges & Resampling
### OIM 7502 — pandas Series, Part 4 of 9

### Generating a range of dates
`pd.date_range` builds a `DatetimeIndex` directly, useful whenever you need a calendar skeleton rather than dates read from a file.

### Resampling
Resampling changes the *frequency* of time-indexed data, and always needs an aggregation function paired with it (`.last()`, `.mean()`, `.sum()`, etc.) — resampling alone just defines the new buckets; you still have to say how to combine what falls into each one.

**A version note:** pandas renamed several frequency aliases — `'M'` (month-end) is now `'ME'`, and `'BM'` (business month-end) is now `'BME'`. Older code and tutorials online will use the old spelling and error on current pandas.

### Be careful with resampling
Resampling to a frequency that starts *before* your data does can silently introduce a leading `NaN` — always sanity-check the result rather than assuming it matches your input 1:1.
