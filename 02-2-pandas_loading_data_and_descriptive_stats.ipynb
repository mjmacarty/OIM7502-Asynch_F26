Now let's move from toy data to something real: daily price data for GBTC (Grayscale Bitcoin Trust).

### Percent change and log returns
`.pct_change()` gives you simple period-over-period returns. Log returns (`np.log(price).diff()`) are the version you'll see more often in finance — they're additive across time, which simple returns aren't.

### Descriptive statistics

### Cumulative sum
Summing log returns cumulatively approximates cumulative growth over the period — this is why log returns are convenient for this kind of analysis.

### Correlation across all numeric columns
`.corr()` gives you a full correlation matrix in one call. Unsurprisingly for OHLC price data, Open/High/Low/Close are all highly correlated with each other — Volume is the outlier.
