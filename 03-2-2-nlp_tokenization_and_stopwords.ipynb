# NLP: Tokenization & Stopword Removal
### OIM 7502 — NLP Series, Part 2 of 5

Raw scraped text is one long string. Most NLP tasks need it broken into individual words (**tokens**) first — and often, with common low-information words (**stopwords**: "the", "and", "of", etc.) filtered out.

**A flag before we start:** stopword removal is genuinely useful for some tasks (word clouds, raw frequency counts) and genuinely harmful for others (sentiment analysis) — that distinction is important enough that Part 4 of this series is entirely about it. For now, we're building the tool; we'll be more careful about when to use it later.

### Tokenizing
A `RegexpTokenizer` splits text into tokens matching a pattern — this one keeps words (including contractions like "don't") and drops punctuation.

### Removing stopwords
NLTK ships a built-in stopword list per language.

**Worth noticing:** NLTK's stopword list includes words like `not`, `no`, `nor`, and `very` — words that carry real meaning for sentiment and emphasis, even though they're grammatically "function words." That's fine for a word cloud or a raw word-frequency count. It is *not* fine for sentiment analysis, which is exactly why Part 4 exists.
