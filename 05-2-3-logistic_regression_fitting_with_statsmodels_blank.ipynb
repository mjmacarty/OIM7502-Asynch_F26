# Logistic Regression: Fitting the Model
### OIM 7502 — Logistic Regression Series, Part 3 of 4

With the design matrix built (Part 2), fitting is one line — `sm.Logit` instead of `sm.OLS`. Same `add_constant`/`.fit()` pattern as linear regression.

### Reading the p-values
`P>|z|` past 0.05 means we can't confidently say that predictor matters, holding the others constant. Here, `children`, `educ`, and most of the individual occupation dummies (`occ2`-`occ4`, all six `hocc` columns) come back statistically insignificant — which is a legitimate reason to simplify the model, not a bug in the data.

### Refitting a leaner model
Drop the insignificant predictors and refit. `occ5`/`occ6` survive because they *were* significant in the full model — occupation categories 5 and 6 (relative to the baseline, occupation 1) do appear to matter.

The pseudo R-squared barely moves (0.1365 → 0.1337) even after dropping 10 predictors — a strong sign those variables genuinely weren't contributing much, and the simpler model is preferable. Every remaining predictor is now significant at p < 0.05.
