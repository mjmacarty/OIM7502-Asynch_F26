An n-gram is a sequence of *n* consecutive words. Looking at n-grams instead of single words surfaces phrases, not just individual word frequency — "amortization table" is far more informative as a pair than "amortization" and "table" counted separately.

```python
from nltk import ngrams, FreqDist

# continuing from Part 2's filtered_text
```

```python
freqs = dict()
for n in [1, 2, 3, 4]:
    freqs[n] = FreqDist(ngrams(filtered_text.split(), n))
```

### Most common single words

```python
freqs[1].most_common(10)
```

### Most common bigrams (2-word phrases)

```python
freqs[2].most_common(10)
```

Notice how much more informative the bigrams are — single-word frequency mixes together every use of a common word, while bigrams start to reveal actual recurring phrases and topics in the source text.
