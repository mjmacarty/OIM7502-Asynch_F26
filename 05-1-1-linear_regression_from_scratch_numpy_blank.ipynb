# Simple Linear Regression, From Scratch in NumPy
### OIM 7502 — Linear Regression Series, Part 1 of 5

Before we reach for `statsmodels` or `sklearn`, it's worth seeing that linear regression isn't magic — it's a well-defined calculation, and NumPy can do it directly with no modeling library at all. This gives you a feel for what those libraries are actually doing under the hood.

### Generate two correlated variables
We'll build `y` as a noisy function of `x`, so we know there's a real linear relationship to recover.

### Step 1: How correlated are they?
`np.corrcoef` returns a correlation matrix — the off-diagonal values are the correlation between the two variables (it's symmetric, so both off-diagonal entries are the same number).

### Step 2: Fit a line with `np.polyfit`
`np.polyfit(x, y, deg=1)` fits a degree-1 polynomial — i.e., a straight line — and returns `[slope, intercept]`. This is the fastest way to get a regression line out of NumPy directly.

### Step 3: The more explicit method — `np.linalg.lstsq`
`polyfit` is convenient, but it's worth seeing the more general "least squares" method it's built on, since this is the same underlying approach `statsmodels` and `sklearn` use. It requires a small trick: adding a column of 1s alongside `x` so the model has an intercept term to solve for (without it, the fit would be forced through the origin).

The `rcond` argument controls how the solver handles numerically tiny values — passing `None` tells NumPy to use its current default behavior rather than an older, deprecated one.

Same answer as `polyfit`, as it should be — it's the same underlying math, just spelled out more explicitly. `polyfit` is doing this same `vstack`-and-solve internally; it just hides the mechanics from you.

**Where this goes next:** what NumPy *doesn't* give you here is anything about how good this fit is — no R², no p-values, no confidence intervals. That's exactly the gap `statsmodels` fills, which is where we're headed next.
